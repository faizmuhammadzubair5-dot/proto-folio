```python
# Beginner Python practice: number analyzer

number = int(input("Enter a number: "))

if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")
else:
    print("The number is zero.")

if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

```python
# Beginner Python practice: password checker

password = input("Create a password: ")

if len(password) < 8:
    print("Your password is too short.")
elif password.isalpha():
    print("Add at least one number or special character.")
elif password.isnumeric():
    print("Add at least one letter.")
else:
    print("Your password is acceptable.")

if password == "123e45678":
    print("Attendance marked: Present")
else:
    print("Invalid password. Attendance not marked.")
max_attempts = 3
attendance_marked = False

for attempt in range(1, max_attempts + 1):
    entered_password = input(f"Enter password (attempt {attempt}/{max_attempts}): ")

    if entered_password == password:
        print("Attendance marked: Present")
        attendance_marked = True
        break
    else:
        remaining = max_attempts - attempt
        if remaining:
            print(f"Wrong password. Try again. Attempts left: {remaining}")
        else:
            print("Attendance marked: Absent")
            print("ALERT: Maximum password attempts exceeded!")
```
